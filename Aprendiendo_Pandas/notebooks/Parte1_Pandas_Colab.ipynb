## ✅ Soluciones (revisión en clase)
> 💡 Colapsa este título (flecha a la izquierda) para ocultar las soluciones hasta la revisión.

```python
def ejercicio_1_alumnos_aprobados(df):
    # Máscara booleana: True donde nota_final >= 5
    return df[df["nota_final"] >= 5]

def ejercicio_2_promedio_por_curso(df):
    # Agrupar por curso y promediar la nota final
    return df.groupby("curso")["nota_final"].mean()

def ejercicio_3_alumno_mejor_nota(df):
    # idxmax() devuelve el índice de la fila con el máximo; loc trae el nombre
    return df.loc[df["nota_final"].idxmax(), "nombre"]

def ejercicio_4_contar_por_ciudad(df):
    # value_counts() omite los NaN; con dropna=False también los cuenta
    return df["ciudad"].value_counts()

def ejercicio_5_nueva_columna_estado(df):
    df = df.copy()   # no modificar el DataFrame original
    df["estado"] = df["nota_final"].apply(lambda n: "Aprobado" if n >= 5 else "Suspenso")
    return df
```

```python
print("1) Aprobados:", len(ejercicio_1_alumnos_aprobados(df)), "de", len(df))
print("\n2) Promedio por curso:\n", ejercicio_2_promedio_por_curso(df).round(2))
print("\n3) Mejor nota:", ejercicio_3_alumno_mejor_nota(df))
print("\n4) Alumnos por ciudad:\n", ejercicio_4_contar_por_ciudad(df))
print("\n5) Estado:\n", ejercicio_5_nueva_columna_estado(df)["estado"].value_counts())
```
